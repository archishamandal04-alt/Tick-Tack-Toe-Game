```python
board= [" ", " ", " ", " "," ", " ", " ", " "," "]

win= [[0,1,2], [3,4,5],[6,7,8],[0,3,6],[1,4,7],[2,5,8],[0,4,8],[2,4,6]]

def showboard():
  print(board[0], "|", board[1], "|", board[2])
  print("--+--+--")
  print(board[3], "|", board[4], "|", board[5])
  print("--+--+--")
  print(board[6], "|", board[7], "|", board[8])


def checkwinner():
  for a , b, c in win:
    if board[a]==board[b]==board[c] and board[a]!=" ":
      return True
  return False


player= "X"

for turn in range(9):
  showboard()

  position =int(input("Player " + player + " entre position 1-9 "))
  if board[position - 1] == " ":
    board[position - 1]= player
  else:
    print("Sorry! the position is already taken")
    continue


  if checkwinner():
    showboard()
    print("Player", player, "wins, Congratulations!")
    break


  if player== "X":
    player= "O"
  else:
    player="X"

else:
  showboard()
  print("it is a draw")
```
